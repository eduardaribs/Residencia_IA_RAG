## Teste 1 - Chunk fixo de 200 caracteres, sem overlap

```python
from langchain_text_splitters import CharacterTextSplitter
from pathlib import Path
```

```python
arquivo = Path("markdown/bioetica_e_ia.md")

texto = arquivo.read_text(encoding="utf-8")

print("Quantidade de caracteres:", len(texto))
```

```python
splitter = CharacterTextSplitter(
    separator="",
    chunk_size=200,
    chunk_overlap=0
)
```

```python
chunks = splitter.split_text(texto)

print("Quantidade de chunks:", len(chunks))
```

```python
for i, chunk in enumerate(chunks[:5], start=1):
    print("=" * 60)
    print(f"CHUNK {i}")
    print(chunk)
    print("Tamanho:", len(chunk))
```

```python
tamanhos = [len(chunk) for chunk in chunks]

print("Número total de chunks:", len(chunks))
print("Tamanho médio:", sum(tamanhos) / len(tamanhos))
print("Tamanho mínimo:", min(tamanhos))
print("Tamanho máximo:", max(tamanhos))
print("Overlap:", 0)
```

```python
splitter_teste_2 = CharacterTextSplitter(
    separator="",
    chunk_size=500,
    chunk_overlap=0
)

chunks_teste_2 = splitter_teste_2.split_text(texto)

tamanhos_teste_2 = [len(chunk) for chunk in chunks_teste_2]

print("Número total de chunks:", len(chunks_teste_2))
print("Tamanho médio:", sum(tamanhos_teste_2) / len(tamanhos_teste_2))
print("Tamanho mínimo:", min(tamanhos_teste_2))
print("Tamanho máximo:", max(tamanhos_teste_2))
print("Overlap:", 0)
```

Em contrapartida, o overlap gera redundância de conteúdo e tende a aumentar
a quantidade de informação processada e armazenada.

## Teste 7 - Chunking por Parágrafos

Nesta estratégia, o documento é dividido utilizando as quebras de parágrafo
como separadores, buscando preservar unidades textuais completas e reduzir
cortes arbitrários no conteúdo.

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter_teste_7 = RecursiveCharacterTextSplitter(
    separators=["\n\n"],
    chunk_size=1000,
    chunk_overlap=0,
    length_function=len
)

chunks_teste_7 = splitter_teste_7.split_text(texto)

print("Número total de chunks:", len(chunks_teste_7))
```

```python
for i, chunk in enumerate(chunks_teste_7[:5], start=1):
    print("=" * 60)
    print(f"CHUNK {i}")
    print(chunk)
    print("\nTamanho:", len(chunk))
```

```python
tamanhos_teste_7 = [len(chunk) for chunk in chunks_teste_7]

print("Número total de chunks:", len(chunks_teste_7))
print("Tamanho médio:", round(sum(tamanhos_teste_7) / len(tamanhos_teste_7), 2))
print("Tamanho mínimo:", min(tamanhos_teste_7))
print("Tamanho máximo:", max(tamanhos_teste_7))
```

### Observação do Teste 7

A estratégia baseada em parágrafos preservou melhor as unidades textuais naturais
do documento e evitou cortes arbitrários no meio das palavras.

Entretanto, alguns parágrafos possuem mais de 1000 caracteres, fazendo com que
determinados chunks ultrapassem o valor definido em `chunk_size`.

Esse comportamento ocorre porque, neste experimento, foi utilizado apenas
`\n\n` como separador, priorizando a preservação integral dos parágrafos.

O resultado evidencia um dos trade-offs do chunking estrutural: maior coerência
semântica pode resultar em chunks de tamanhos menos uniformes.

```python
tamanhos_teste_7 = [len(chunk) for chunk in chunks_teste_7]

print("Número total de chunks:", len(chunks_teste_7))
print("Tamanho médio:", round(sum(tamanhos_teste_7) / len(tamanhos_teste_7), 2))
print("Tamanho mínimo:", min(tamanhos_teste_7))
print("Tamanho máximo:", max(tamanhos_teste_7))
```

Nesta estratégia, o documento é dividido com base nos headings Markdown,

## Comparação preliminar das estratégias

Os experimentos demonstraram que o tamanho do chunk influencia diretamente
a quantidade de fragmentos gerados. Entre os testes de tamanho fixo sem
overlap, o uso de 200 caracteres produziu 222 chunks, enquanto 2000
caracteres resultaram em apenas 23 chunks.

A introdução de overlap aumentou a quantidade de chunks. Para chunks de
500 caracteres, foram obtidos 89 chunks sem overlap, 99 com overlap de
50 caracteres e 148 com overlap de 200 caracteres. Esse resultado evidencia
o aumento da redundância causado pela sobreposição.

As estratégias estruturais apresentaram comportamentos diferentes. O
chunking por parágrafos gerou 57 chunks e preservou melhor unidades textuais,
enquanto o agrupamento de três sentenças produziu 86 chunks com tamanhos
variáveis.

O RecursiveCharacterTextSplitter apresentou 65 chunks, buscando equilibrar
o limite de tamanho com a preservação da estrutura textual.

Por fim, o MarkdownHeaderTextSplitter gerou apenas 19 chunks, apresentando
a maior média de tamanho entre as estratégias. Apesar de preservar headings
e metadados estruturais, algumas seções resultaram em chunks muito extensos.

```python
def obter_texto_chunk(chunk):
    if hasattr(chunk, "page_content"):
        return chunk.page_content
    return str(chunk)


def obter_metadata_chunk(chunk):
    if hasattr(chunk, "metadata"):
        return chunk.metadata
    return {}
```

```python
resultados_embeddings = []

document_id = "doc01"
document_name = "bioetica_e_ia.md"

for teste_id, config in experimentos.items():

    print("=" * 60)
    print(f"PROCESSANDO TESTE {teste_id}")
    print(f"Estratégia: {config['strategy']}")
    print(f"Quantidade de chunks: {len(config['chunks'])}")

    for indice, chunk in enumerate(config["chunks"], start=1):

        texto_chunk = obter_texto_chunk(chunk)
        metadata = obter_metadata_chunk(chunk)

        try:
            embedding = get_embedding(texto_chunk)

            resultado = {
                "chunk_id": f"{document_id}_test{teste_id:02d}_chunk{indice:03d}",
                "document_id": document_id,
                "document_name": document_name,
                "test_id": teste_id,
                "strategy": config["strategy"],
                "chunk_size": config["chunk_size"],
                "chunk_overlap": config["chunk_overlap"],
                "text": texto_chunk,
                "embedding": embedding,
                "metadata": metadata
            }

            resultados_embeddings.append(resultado)

            print(
                f"Teste {teste_id} | "
                f"Chunk {indice}/{len(config['chunks'])} | "
                f"Embedding: {len(embedding)} dimensões"
            )

            # Pequena pausa para evitar muitas chamadas seguidas
            time.sleep(0.2)

        except Exception as erro:
            print(
                f"ERRO no Teste {teste_id}, "
                f"Chunk {indice}: {erro}"
            )

print("\nPROCESSAMENTO FINALIZADO")
print("Total de registros:", len(resultados_embeddings))
```

# Análise dos Experimentos

## 1. Qual estratégia gerou mais chunks?

Entre os experimentos realizados, a estratégia com maior número de chunks foi o
Teste 1, com chunk_size de 200 caracteres e sem overlap.

Esse comportamento era esperado, pois quanto menor o tamanho máximo permitido
para cada chunk, maior é a quantidade de fragmentos necessários para representar
o documento completo.

## 2. Qual estratégia gerou menos chunks?

A estratégia que gerou menos chunks foi o MarkdownHeaderTextSplitter.

Essa estratégia utiliza a estrutura de headings do documento para definir as
divisões, produzindo menos chunks, porém com tamanhos significativamente maiores
e mais variáveis.

## 3. Como o tamanho dos chunks variou?

As estratégias baseadas em tamanho fixo produziram chunks mais uniformes.

Nos testes com 200, 500, 1000 e 2000 caracteres, o tamanho dos chunks se manteve
próximo ao limite configurado.

Já as estratégias baseadas em parágrafos, sentenças e headings apresentaram maior
variação, pois priorizam unidades estruturais ou semânticas do texto em vez de
limites rígidos de tamanho.

## 4. Qual estratégia preservou melhor a estrutura dos documentos?

O MarkdownHeaderTextSplitter apresentou a melhor preservação da estrutura
documental, pois manteve informações associadas aos headings e seções.

O chunking por parágrafos também preservou melhor a estrutura textual do que os
métodos puramente baseados em caracteres.

## 5. Como tabelas foram tratadas?

As tabelas já haviam sido convertidas para a representação Markdown durante a
etapa PDF → Markdown.

Nos chunkings baseados em caracteres, existe risco de fragmentação das linhas e
da estrutura da tabela.

Estratégias estruturais, especialmente Markdown e Recursive, tendem a apresentar
maior possibilidade de preservação, dependendo da posição e do tamanho da tabela.

## 6. Como imagens foram tratadas?

Durante a conversão para Markdown, algumas imagens foram representadas por
marcadores como:

`<!-- image -->`

Isso indica que a imagem em si não foi incorporada como conteúdo visual no
Markdown utilizado pelos experimentos.

Assim, o embedding textual não representa diretamente o conteúdo visual da
imagem.

## 7. Quais informações foram perdidas durante a conversão PDF → Markdown?

Foi possível observar perda parcial da estrutura visual original do PDF.

Entre os elementos afetados estão layout de páginas, posicionamento de imagens,
colunas, espaçamento e parte da formatação visual.

Também foram observadas palavras fragmentadas por hifenização proveniente do PDF.

## 8. O chunking por caracteres fragmentou conceitos ou estruturas importantes?

Sim.

Os testes de chunking fixo apresentaram cortes no meio de palavras, frases e
parágrafos.

Isso demonstra que a divisão baseada exclusivamente em número de caracteres pode
prejudicar a preservação do contexto semântico.

## 9. O chunking por parágrafo produziu chunks muito grandes?

Em alguns casos, sim.

Como determinados parágrafos possuíam mais caracteres do que o tamanho definido,
alguns chunks ultrapassaram o valor configurado.

Esse resultado mostra o equilíbrio necessário entre preservação semântica e
controle do tamanho dos chunks.

## 10. O chunking por sentença conseguiu preservar melhor o contexto?

O agrupamento de três sentenças reduziu os cortes arbitrários encontrados nas
estratégias baseadas em caracteres.

Essa abordagem preservou unidades completas de sentido, embora tenha produzido
chunks de tamanhos variados.

## 11. O Recursive Splitter apresentou vantagens?

Sim.

O RecursiveCharacterTextSplitter apresentou um equilíbrio entre controle de
tamanho e preservação da estrutura textual.

A estratégia tenta inicialmente separar por parágrafos, depois por linhas,
espaços e, por último, caracteres.

Isso reduz a fragmentação arbitrária quando comparado ao chunking fixo.

## 12. O Markdown Splitter conseguiu preservar a estrutura semântica?

Sim.

O MarkdownHeaderTextSplitter conseguiu associar conteúdo aos headings e preservar
metadados relacionados às seções.

Essa característica pode ser especialmente útil em sistemas de RAG, pois permite
recuperar não apenas o texto, mas também informações sobre a seção de origem.

## 13. Qual estratégia parece mais adequada para um sistema de RAG?

Entre as estratégias avaliadas, o RecursiveCharacterTextSplitter apresenta uma
boa relação entre tamanho dos chunks e preservação do contexto.

O MarkdownHeaderTextSplitter também apresentou vantagens importantes na
preservação da estrutura semântica.

Uma abordagem híbrida, combinando headings Markdown com uma subdivisão recursiva
das seções muito grandes, pode ser uma alternativa adequada para RAG.

## 14. Quais estratégias devem ser descartadas?

Os chunkings fixos muito pequenos, especialmente o de 200 caracteres, tendem a
fragmentar excessivamente o conteúdo e gerar grande quantidade de embeddings.

O overlap muito alto também pode ser menos interessante devido à grande
redundância e ao aumento do custo de processamento e armazenamento.

## 15. Quais estratégias devem ser utilizadas nos próximos experimentos?

As estratégias mais interessantes para os próximos testes são:

- RecursiveCharacterTextSplitter;
- MarkdownHeaderTextSplitter;
- chunking por parágrafos;
- eventualmente chunk fixo intermediário com overlap leve, como 500 caracteres
  com overlap 50, para comparação.

Essas estratégias apresentam melhor equilíbrio entre preservação de contexto,
estrutura semântica, quantidade de chunks e custo de processamento.

```python
from pathlib import Path
from pypdf import PdfReader
import gc

arquivo_pdf = Path("artigos/gpt3_language_models.pdf")
arquivo_saida = Path("markdown/gpt3_language_models.md")

# Processar apenas 2 páginas por vez
TAMANHO_BLOCO = 2

reader = PdfReader(str(arquivo_pdf))
total_paginas = len(reader.pages)

print("Arquivo:", arquivo_pdf.name)
print("Total de páginas:", total_paginas)
print("=" * 60)

partes_markdown = []

for inicio in range(1, total_paginas + 1, TAMANHO_BLOCO):
    fim = min(inicio + TAMANHO_BLOCO - 1, total_paginas)

    print(
        f"⏳ Processando páginas {inicio}-{fim} de {total_paginas}...",
        flush=True
    )

    try:
        resultado = converter.convert(
            arquivo_pdf,
            page_range=(inicio, fim)
        )

        markdown_parcial = resultado.document.export_to_markdown()

        partes_markdown.append(markdown_parcial)

        print(
            f"✅ Páginas {inicio}-{fim} concluídas!",
            flush=True
        )

        del resultado
        del markdown_parcial
        gc.collect()

    except Exception as erro:
        print(
            f"❌ ERRO nas páginas {inicio}-{fim}: {erro}",
            flush=True
        )
        gc.collect()

markdown_completo = "\n\n".join(partes_markdown)

arquivo_saida.write_text(
    markdown_completo,
    encoding="utf-8"
)

print("\n" + "=" * 60)
print("✅ CONVERSÃO FINALIZADA")
print("Arquivo criado:", arquivo_saida)
print("Tamanho:", arquivo_saida.stat().st_size, "bytes")
```

```python
import json
from pathlib import Path

from langchain_text_splitters import (
    RecursiveCharacterTextSplitter,
    MarkdownHeaderTextSplitter,
)

# ============================================================
# PASTAS
# ============================================================

PASTA_MARKDOWN = Path("markdown")
PASTA_RESULTS = Path("results")

PASTA_RESULTS.mkdir(
    parents=True,
    exist_ok=True
)


# ============================================================
# SPLITTERS
# ============================================================

# 1. Recursive
splitter_recursive = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=100,
    separators=[
        "\n\n",
        "\n",
        " ",
        ""
    ],
    length_function=len
)

# 2. Markdown por headings
headers_to_split_on = [
    ("#", "titulo_nivel_1"),
    ("##", "titulo_nivel_2"),
    ("###", "titulo_nivel_3"),
]

splitter_markdown = MarkdownHeaderTextSplitter(
    headers_to_split_on=headers_to_split_on
)


# ============================================================
# FUNÇÃO PARA CHUNKING POR PARÁGRAFOS
# ============================================================

def chunking_paragrafos(texto):
    """
    Divide o texto preservando parágrafos como unidades.
    """
    paragrafos = [
        p.strip()
        for p in texto.split("\n\n")
        if p.strip()
    ]

    return paragrafos


# ============================================================
# FUNÇÃO PARA SALVAR JSON
# ============================================================

def salvar_json(caminho, dados):
    caminho.parent.mkdir(
        parents=True,
        exist_ok=True
    )

    with open(
        caminho,
        "w",
        encoding="utf-8"
    ) as arquivo:

        json.dump(
            dados,
            arquivo,
            ensure_ascii=False,
            indent=2
        )


# ============================================================
# PROCESSAR TODOS OS MARKDOWNS
# ============================================================

arquivos_md = sorted(
    PASTA_MARKDOWN.glob("*.md")
)

print(
    "Quantidade de documentos:",
    len(arquivos_md)
)

print("=" * 70)


for indice_doc, arquivo_md in enumerate(
    arquivos_md,
    start=1
):

    document_id = f"doc{indice_doc:02d}"
    document_name = arquivo_md.name
    document_slug = arquivo_md.stem

    print(
        f"\n[{indice_doc}/{len(arquivos_md)}] "
        f"Processando: {document_name}"
    )

    texto = arquivo_md.read_text(
        encoding="utf-8"
    )

    pasta_documento = (
        PASTA_RESULTS / document_slug
    )

    # ========================================================
    # ESTRATÉGIA 1 - RECURSIVE
    # ========================================================

    chunks_recursive = (
        splitter_recursive.split_text(texto)
    )

    dados_recursive = []

    for indice_chunk, chunk in enumerate(
        chunks_recursive,
        start=1
    ):

        dados_recursive.append({
            "chunk_id": (
                f"{document_id}_recursive_"
                f"chunk{indice_chunk:03d}"
            ),
            "document_id": document_id,
            "document_name": document_name,
            "strategy": "recursive",
            "chunk_size": 1000,
            "chunk_overlap": 100,
            "text": chunk,
            "metadata": {}
        })

    salvar_json(
        pasta_documento
        / "recursive"
        / "chunks.json",
        dados_recursive
    )

    print(
        f"  ✅ Recursive: "
        f"{len(chunks_recursive)} chunks"
    )


    # ========================================================
    # ESTRATÉGIA 2 - PARÁGRAFOS
    # ========================================================

    chunks_paragrafos = (
        chunking_paragrafos(texto)
    )

    dados_paragrafos = []

    for indice_chunk, chunk in enumerate(
        chunks_paragrafos,
        start=1
    ):

        dados_paragrafos.append({
            "chunk_id": (
                f"{document_id}_paragraph_"
                f"chunk{indice_chunk:03d}"
            ),
            "document_id": document_id,
            "document_name": document_name,
            "strategy": "paragraph",
            "chunk_size": None,
            "chunk_overlap": 0,
            "text": chunk,
            "metadata": {}
        })

    salvar_json(
        pasta_documento
        / "paragraph"
        / "chunks.json",
        dados_paragrafos
    )

    print(
        f"  ✅ Paragraph: "
        f"{len(chunks_paragrafos)} chunks"
    )


    # ========================================================
    # ESTRATÉGIA 3 - MARKDOWN HEADINGS
    # ========================================================

    chunks_markdown = (
        splitter_markdown.split_text(texto)
    )

    dados_markdown = []

    for indice_chunk, chunk in enumerate(
        chunks_markdown,
        start=1
    ):

        dados_markdown.append({
            "chunk_id": (
                f"{document_id}_markdown_"
                f"chunk{indice_chunk:03d}"
            ),
            "document_id": document_id,
            "document_name": document_name,
            "strategy": "markdown",
            "chunk_size": None,
            "chunk_overlap": 0,
            "text": chunk.page_content,
            "metadata": chunk.metadata
        })

    salvar_json(
        pasta_documento
        / "markdown"
        / "chunks.json",
        dados_markdown
    )

    print(
        f"  ✅ Markdown: "
        f"{len(chunks_markdown)} chunks"
    )


print("\n" + "=" * 70)
print("🎉 PROCESSAMENTO FINALIZADO!")
```

## Busca semântica nos chunks

```python
import numpy as np

def similaridade_cosseno(a, b):
    a = np.array(a)
    b = np.array(b)

    return np.dot(a, b) / (
        np.linalg.norm(a) * np.linalg.norm(b)
    )

print("✅ Função de similaridade criada!")
```

```python
import json
from pathlib import Path

def buscar_top_k(estrategia, embedding_query, top_k=5):
    resultados = []

    arquivos = sorted(
        Path("results").glob(
            f"*/{estrategia}/chunks_embeddings.json"
        )
    )

    for arquivo in arquivos:
        with open(arquivo, "r", encoding="utf-8") as f:
            chunks = json.load(f)

        for item in chunks:
            score = similaridade_cosseno(
                embedding_query,
                item["embedding"]
            )

            resultados.append({
                "estrategia": estrategia,
                "documento": item["document_name"],
                "chunk_id": item["chunk_id"],
                "score": float(score),
                "texto": item["text"]
            })

    resultados = sorted(
        resultados,
        key=lambda x: x["score"],
        reverse=True
    )

    return resultados[:top_k]
```

## Análise comparativa das estratégias de chunking

Foram realizadas três consultas semânticas relacionadas a diferentes temas presentes no corpus: mecanismos de atenção, alinhamento de modelos por aprendizado por reforço com feedback humano (RLHF) e recuperação aumentada por geração (RAG). Para cada consulta, foram comparadas as estratégias Recursive, Paragraph e Markdown, considerando os três chunks com maior similaridade semântica.

Os resultados demonstram que nenhuma estratégia apresentou desempenho superior em todos os cenários. Na consulta relacionada aos mecanismos de atenção, o documento *Attention Is All You Need* não ocupou a primeira posição em nenhuma estratégia, embora tenha sido recuperado entre os três primeiros resultados pelas estratégias Paragraph e Markdown.

Na consulta sobre RLHF e alinhamento de modelos, a estratégia Paragraph apresentou um resultado particularmente coerente, recuperando *InstructGPT* na primeira posição. As demais estratégias também recuperaram documentos diretamente relacionados ao tema, especialmente *InstructGPT* e o relatório técnico do GPT-4.

Para a consulta relacionada à recuperação de documentos externos, as três estratégias apresentaram comportamento consistente, recuperando *Retrieval-Augmented Generation* entre os resultados mais relevantes. Nesse caso, Recursive e Paragraph apresentaram scores de similaridade muito próximos na primeira posição.

Esses resultados indicam que o valor numérico da similaridade não deve ser interpretado isoladamente como medida absoluta da qualidade da recuperação. A relevância semântica do conteúdo recuperado também precisa ser considerada. Além disso, diferentes formas de segmentação modificam o contexto representado pelos embeddings e, consequentemente, a posição dos documentos nos resultados da busca.

De maneira geral, a estratégia Paragraph apresentou bons resultados em consultas específicas, enquanto Recursive demonstrou maior regularidade no tamanho dos chunks e resultados consistentes de recuperação. Markdown preservou melhor a estrutura documental, porém produziu segmentos maiores e exigiu subdivisão adicional em alguns documentos para respeitar os limites do modelo de embeddings.
